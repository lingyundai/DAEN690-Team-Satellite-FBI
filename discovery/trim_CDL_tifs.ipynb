CDL tif files downloaded from the USDA site cover the entirety of the US. This code was used to trim CDL tif images to just the Huron County, Michigan area. This trimming was performed for 5 CDL tif files 2019 - 2023.

```python
# AWS API related
import boto3
from botocore import UNSIGNED
from botocore.config import Config
# Work with TIFF format files
import rasterio
# Visualization
import matplotlib.pyplot as plt
# Data manipulation
import numpy as np
# System
import os
from PIL import Image
from rasterio.windows import Window
```

# 2019

```python
#Load Images
CDL_2019_tif_path = "2019_30m_cdls.tif"
CDL_2019_cropped_tif_path = "2019_30m_cdls_cropped.tif"
```

```python
#(Optional Step) count number of pixels in tif file
with rasterio.open(CDL_2019_tif_path) as src:
    # Get the dimensions (width, height)
    width = src.width
    height = src.height
    
    print(f"Raster dimensions: {width} columns (width), {height} rows (height)")
```

```python
# Define the trim window (coor
row_start, row_end = 25250, 26990  # Crop between these rows
col_start, col_end = 111500, 114500  # Crop between these columns
```

```python
# Open the original TIFF file
with rasterio.open(CDL_2019_tif_path) as src:
    # Define a window with the chosen crop dimensions
    window = Window(col_start, row_start, col_end - col_start, row_end - row_start)

    # Read and crop the image using the window
    cropped_data = src.read(window=window)

    # Update metadata to match the new cropped size
    new_meta = src.meta.copy()
    new_meta.update({
        "height": window.height,
        "width": window.width,
        "transform": src.window_transform(window)
    })

    # Save the cropped image
    with rasterio.open(CDL_2019_cropped_tif_path, "w", **new_meta) as dest:
        dest.write(cropped_data)
        
print(f"Cropped TIFF saved as: {CDL_2019_cropped_tif_path}")
```

This part is optional. Here, we convert the TIF to PNG and take a look to make sure we cropped it correctly

```python
# Define input and output file paths
tif_path = "2019_30m_cdls_cropped.tif"
png_path = "converted_image_2019.png"
```

```python
# Open the TIFF file
with rasterio.open(tif_path) as src:
    image_array = src.read(1)  # Read the single available band
```

```python
# Convert to PIL image and save as PNG
image = Image.fromarray(image_array)
image.save(png_path)
print(f"Saved PNG file as: {png_path}")
```
